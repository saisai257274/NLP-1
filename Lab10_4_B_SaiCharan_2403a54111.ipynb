```python
import gensim.downloader as api
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
```

```python
model = api.load("glove-wiki-gigaword-100")

print("Vocabulary Size:", len(model.index_to_key))

print("Vector for 'king':\n", model["king"])
```

```python
words = [
    # Animals
    "dog", "cat", "lion", "tiger", "elephant", "wolf", "monkey",

    # Fruits
    "apple", "banana", "mango", "orange", "grape", "pineapple",

    # Countries
    "india", "china", "america", "japan", "germany", "france",

    # Technology
    "computer", "laptop", "mobile", "internet", "software", "hardware",

    # Vehicles
    "car", "bus", "train", "bike", "truck", "airplane"
]

vectors = np.array([model[word] for word in words])

print("Total words selected:", len(words))
print("Vector shape:", vectors.shape)
print(vectors[0])
```

```python
tsne = TSNE(n_components=2, random_state=42, perplexity=5)
reduced_vectors = tsne.fit_transform(vectors)

print("Shape after reduction:", reduced_vectors.shape)
```

```python
plt.figure(figsize=(12,8))
plt.scatter(reduced_vectors[:,0], reduced_vectors[:,1])

for i, word in enumerate(words):
    plt.annotate(word, (reduced_vectors[i,0], reduced_vectors[i,1]))

plt.title("t-SNE Visualization of Word Embeddings")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.grid(True)
plt.show()
```
