# Task
Perform topic modeling using Latent Dirichlet Allocation (LDA) on a corpus of news headlines. The process involves creating an Excel-like DataFrame for the headlines, cleaning and preprocessing the text, generating a Bag of Words representation, discovering latent topics, extracting key words for each topic, and finally assigning these topics back to the original headlines.

## Prepare and Load Data

### Subtask:
Create a corpus of news headlines and load them into a pandas DataFrame to simulate an Excel-like structure.


**Reasoning**:
I will import the pandas library and create a DataFrame containing a variety of news headlines as specified in the instructions.



```python
import pandas as pd

# Create a list of news headlines across different topics
headlines = [
    "New advancements in AI chips are set to change the industry",
    "Global leaders meet to discuss climate change policies",
    "Recent breakthroughs in cancer research show promising results",
    "The underdog team wins the national championship in a stunning upset",
    "Smartphone manufacturers announce next-generation foldable devices",
    "Health officials urge caution as flu season approaches",
    "Major political shift expected after recent election results",
    "New stadium construction projects boost local economy"
]

# Convert the list into a pandas DataFrame with a single column 'Headline'
df_headlines = pd.DataFrame(headlines, columns=['Headline'])

# Display the DataFrame to verify the content
print("Headlines DataFrame:")
df_headlines
```

## Text Preprocessing

### Subtask:
Clean and preprocess the news headlines to prepare them for vectorization.


**Reasoning**:
To prepare the text for topic modeling, I will clean the headlines by converting them to lowercase and removing any non-alphabetic characters (punctuation and numbers). This ensures that the model focuses on the semantic meaning of the words rather than noise. I will also normalize the spacing by tokenizing and rejoining the words.

**Reasoning**:
I will implement the text preprocessing steps by defining a function to clean the headlines—converting them to lowercase, removing non-alphabetic characters using regular expressions, and normalizing whitespace—and then applying this function to create a new 'Cleaned_Headline' column.



```python
import re

def preprocess_text(text):
    # Convert to lowercase and remove all non-alphabetic characters
    cleaned = re.sub(r'[^a-z\s]', '', text.lower())
    # Tokenize and rejoin to handle extra spaces and normalize spacing
    return ' '.join(cleaned.split())

# Create the new column with cleaned text
df_headlines['Cleaned_Headline'] = df_headlines['Headline'].apply(preprocess_text)

# Display the first few rows to verify the transformation
print('Processed Headlines sample:')
print(df_headlines[['Headline', 'Cleaned_Headline']].head())
df_headlines.head()
```

## Generate Bag of Words (BoW)

### Subtask:
Create a Bag of Words representation of the preprocessed headlines and convert it into a structured DataFrame for inspection.


**Reasoning**:
I will import the CountVectorizer from scikit-learn, initialize it with English stop words, and transform the cleaned headlines into a Bag of Words DataFrame for inspection.



```python
from sklearn.feature_extraction.text import CountVectorizer

# Initialize CountVectorizer to remove English stop words
vectorizer = CountVectorizer(stop_words='english')

# Fit and transform the 'Cleaned_Headline' column
bow_matrix = vectorizer.fit_transform(df_headlines['Cleaned_Headline'])

# Convert the sparse matrix to a dense pandas DataFrame
df_bow = pd.DataFrame(bow_matrix.toarray(), columns=vectorizer.get_feature_names_out())

# Display the resulting Bag of Words DataFrame
print("Bag of Words (BoW) Representation:")
df_bow
```

## Apply LDA Modeling

### Subtask:
Apply Latent Dirichlet Allocation (LDA) to the BoW model to discover latent topics.


**Reasoning**:
I will import the LatentDirichletAllocation class, initialize the model with 4 components, fit it to the Bag of Words matrix, and transform the matrix to obtain the topic distributions as requested.



```python
from sklearn.decomposition import LatentDirichletAllocation

# Initialize LDA model with 4 topics
lda_model = LatentDirichletAllocation(n_components=4, random_state=42)

# Fit the model to the Bag of Words matrix
lda_model.fit(bow_matrix)

# Transform the BoW matrix to get topic distributions for each headline
lda_output = lda_model.transform(bow_matrix)

# Confirm the modeling step
print("LDA Model fitted:", lda_model)
print("Shape of lda_output (headlines, topics):", lda_output.shape)
print("\nTopic distributions for the first few headlines:")
print(lda_output[:5])
```

## Identify Topics and Words

### Subtask:
Extract the key words associated with each discovered topic to understand the thematic structure.


**Reasoning**:
I will extract the top 10 words for each topic by accessing the components of the LDA model and mapping the highest-weight word indices back to the feature names from the vectorizer.



```python
import numpy as np

# Retrieve the feature names (words) from the vectorizer
words = vectorizer.get_feature_names_out()

# Iterate through each of the 4 topics to find the top 10 words
for i, topic in enumerate(lda_model.components_):
    # Get the indices of the top 10 words with the highest weights for this topic
    top_words_idx = topic.argsort()[-10:][::-1]
    # Map indices back to words
    top_words = [words[idx] for idx in top_words_idx]

    print(f"Topic {i+1}:")
    print(", ".join(top_words))
    print("-" * 20)
```

## Assign Topics to Documents

### Subtask:
Map the identified topics back to each original headline by finding the dominant topic in the LDA output and adding it as a new column to the DataFrame.


**Reasoning**:
I will map the dominant topic to each headline by identifying the topic index with the highest probability from the LDA results and adding both the topic label and its probability score to the DataFrame.



```python
# Identify the dominant topic for each document (headline)
df_headlines['Topic'] = np.argmax(lda_output, axis=1) + 1

# Extract the probability of the dominant topic
df_headlines['Topic_Probability'] = np.max(lda_output, axis=1)

# Display the final DataFrame with original headlines and their assigned topics
print("Headlines with Assigned Topics:")
df_headlines[['Headline', 'Topic', 'Topic_Probability']]
```
